Import Libraries


```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

Load Dataset

```python
df=pd.read_csv("/content/fake_news_dataset.csv")
print(df.head())
```

Preprocess the Data

```python
# Check for missing values
print(df.isnull().sum())

# Fill missing values with empty string
df = df.fillna('')

# Combine 'author' and 'title' into a single feature
df['content'] = df['author'] + ' ' + df['title']
```

Split Data into Train and Test

```python
# Features and Target
X = df['content']
y = df['label']

# Split data (75% training, 20% testing)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)
```

Text Vectorization (TF-IDF)

```python
# Initialize TF-IDF Vectorizer
vectorizer = TfidfVectorizer()

# Fit and transform the training data
X_train_vec = vectorizer.fit_transform(X_train)

# Transform the testing data
X_test_vec = vectorizer.transform(X_test)
```

Train the Machine Learning Model

```python
# Initialize Logistic Regression model
model = LogisticRegression()

# Train the model
model.fit(X_train_vec, y_train)
```

Evaluate the Model

```python
# Predict on test set
y_pred = model.predict(X_test_vec)

# Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)
print(f"Model Accuracy: {accuracy * 100:.2f}%")

# Confusion Matrix
print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred))

# Classification Report
print("Classification Report:")
print(classification_report(y_test, y_pred))
```

Make a Custom Prediction

```python
# Example text (You can replace this with any news content)
new_input = ["Breaking news! Scientists find a new planet beyond Pluto!"]

# Preprocess and predict
new_input_vec = vectorizer.transform(new_input)
prediction = model.predict(new_input_vec)

# Show result
if prediction[0] == 0:
    print("The news is Real 📰✅")
else:
    print("The news is Fake 🚫📰")
```
